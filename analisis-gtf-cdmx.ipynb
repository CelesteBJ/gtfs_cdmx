```python
import geopandas as gd
rutas=gd.read_file("data/STC_Metro_lineas.kmz")
print(rutas.head())
```

```python
import folium

# Coordenadas del Centro Histórico de la Ciudad de México
mapa=folium.Map(location=(19.4326, -99.1332))
```

```python
# Convertir MULTILINESTRING a array de tuplas (latitud, longitud)
for idx, ruta in rutas.iterrows():
    geom=ruta["geometry"]
    coords=[]
    for line in geom.geoms:  # Iterar sobre cada línea del MULTILINESTRING
        for coord in line.coords:  # Iterar sobre cada coordenada
            coords.append((coord[1], coord[0]))  # Solo lon, lat (ignorar Z)
    folium.PolyLine(coords).add_to(mapa)
```

```python
mapa.save("test/index.html")
```
